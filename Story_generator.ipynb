```python
from google import genai
from google.genai import types
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))
```

```python
MODEL = "models/gemini-2.5-flash-lite"
```

```python
def generate_story(topic, genre, length_words=300):
    system_prompt_content = ("""
You are a creative and professional short story writer.

Your task is to write an engaging, original, and complete short story based
on the topic and genre provided by the user.

Follow these instructions:
- Create a clear beginning, middle, and ending.
- Develop the given topic into a meaningful story.
- Follow the requested genre consistently.
- Create interesting characters, setting, conflict, and emotions where appropriate.
- Use natural storytelling and descriptive language.
- Include dialogue when it improves the story.
- Maintain logical and consistent events throughout the story.
- Keep the story close to the requested word count.
- Do not explain your writing process.
- Do not provide notes, suggestions, or analysis.
- Output only the story.
""")

    user_prompt = f"""
    write a short story.
    topic:{topic}
    Genre:{genre}
    length: approx {length_words}
    """

    response = client.models.generate_content(
        model=MODEL,
        contents=user_prompt,
        config=types.GenerateContentConfig(
            system_instruction=system_prompt_content,
            temperature=1,
            max_output_tokens=500
        )
    )
    print(response.text)
```

### Generate Your Story!

```python

story_topic = input("Enter the story topic: ")
story_genre = input("Enter the story genre: ")
story_length = int(input("Enter the approximate story length in words: "))

generate_story(story_topic, story_genre, story_length)
```
