````python
from google import genai
from google.genai import types
from google.colab import userdata

client = genai.Client(api_key=userdata.get('GEMINI_API_KEY'))
````

````python
knowledge_base="""
Q: How do I join the gym?
A: Open the FitZone app, select a membership plan, enter your details,
choose a payment method, and complete the registration.

Q: What are the gym opening hours?
A: FitZone is open from 5:00 AM to 10:00 PM Monday to Saturday.
Sunday timings may vary.

Q: Can I freeze my membership?
A: Membership freezing may be available depending on your plan.
Contact customer support to check your eligibility.

Q: How can I book a personal trainer?
A: Open the Trainers section in the FitZone app, select an available
trainer, choose a time slot, and confirm your booking.

Q: How can I cancel my membership?
A: Open Account Settings and select Membership. Follow the cancellation
instructions or contact customer support if the option is unavailable.

Q: What should I do if gym equipment is damaged?
A: Report the equipment issue through the Help section of the FitZone app
or inform a gym staff member.

Q: Can beginners follow the workout plans?
A: Yes. FitZone provides beginner-friendly workout plans. Choose a
beginner plan based on your fitness goals.

```python
Q: Can I upgrade my current membership plan?
A: Yes. You can request a membership upgrade through the FitZone app or contact customer support to check the available upgrade options.

Q: Can I downgrade my membership plan?
A: Membership downgrades may be available depending on your current plan. Contact FitZone customer support to check the available options.

Q: Can I transfer my membership to another person?
A: Membership transfers may depend on the membership terms. Contact FitZone customer support to check whether your membership can be transferred.

Q: Can I use my membership at another FitZone branch?
A: Access to other FitZone branches depends on your membership plan. Check your membership details or contact customer support for confirmation.

Q: Is a trial membership available?
A: Trial membership availability may vary. Contact FitZone customer support to check whether a trial membership is currently available.

Q: Can I extend my membership after it expires?
A: Yes, you may be able to renew or extend your membership through the FitZone app or by contacting customer support.

Q: Can I change my membership start date?
A: Membership start date changes may depend on your plan. Contact FitZone customer support for assistance.

Q: How can I update my payment method?
A: Open the Payment or Membership section in the FitZone app and update your available payment method.

Q: What should I do if my membership payment failed?
A: Check your payment details and try the payment again. If the payment continues to fail, contact FitZone customer support.

Q: Can I get a receipt for my membership payment?
A: You can check your payment or membership history in the FitZone app. Contact customer support if you need additional assistance.

Q: Will my membership renew automatically?
A: Automatic renewal depends on your membership plan and payment settings. Check your membership details in the FitZone app.

Q: How can I choose a workout plan?
A: Open the Workout section in the FitZone app and select a workout plan based on your fitness goals and experience level.

Q: Can beginners use FitZone workout plans?
A: Yes. FitZone provides beginner-friendly workout plans that can be selected from the Workout section.

Q: Can I change my workout plan?
A: Yes. You can select a different available workout plan from the Workout section based on your current fitness goals.

Q: Can I track my workout progress?
A: Yes. You can record your workouts and review your progress through the tracking features available in the FitZone app.

Q: Can I change my fitness goal?
A: Yes. You can update your fitness goal through your profile or workout settings in the FitZone app.

Q: Does FitZone provide home workout programs?
A: Home workout programs may be available depending on the programs offered through the FitZone app.

Q: Does FitZone provide muscle-building workout plans?
A: Yes. FitZone may provide workout plans designed for different fitness goals, including muscle building, depending on the available programs.

Q: Does FitZone provide weight-loss workout plans?
A: Yes. FitZone may provide workout plans designed for different fitness goals, including weight management, depending on the available programs.

Q: Can I change my personal trainer?
A: Trainer changes may be possible depending on trainer availability. Contact FitZone customer support for assistance.

Q: Can I cancel my personal training session?
A: You may be able to cancel a personal training session depending on the booking terms. Check your booking or contact customer support.

Q: Can I reschedule my personal training session?
A: Rescheduling may be available depending on trainer availability and the booking terms. Contact FitZone support if you need assistance.

Q: How can I see available personal trainers?
A: Open the Trainers section in the FitZone app to view available trainers and their available time slots.

Q: How do I book a fitness class?
A: Open the Classes section in the FitZone app, select an available class and time slot, and confirm your booking.

Q: Can I cancel a fitness class booking?
A: Class cancellation may be available depending on the booking terms. Open your booking and check the available cancellation option.

Q: Can I reschedule a fitness class?
A: You may be able to reschedule a class if another suitable time slot is available. Check your booking or contact FitZone support.

Q: How can I see upcoming fitness classes?
A: Open the Classes section in the FitZone app to view available and upcoming fitness classes.

Q: Does FitZone provide locker facilities?
A: Locker availability depends on the facilities provided at the FitZone gym location. Contact the gym or customer support for confirmation.

Q: Can I bring a friend to FitZone?
A: Guest access depends on the membership plan and gym policies. Contact FitZone customer support to check whether guest access is available.

"""


system_instruction = f"""

You are a friendly and helpful customer support assistant for a gym and
fitness application called FitZone.

Your job is to help customers with questions about gym memberships,
membership plans, payments, gym timings, workout programs, personal
trainers, fitness classes, equipment, nutrition information, account
issues, bookings, and other gym-related services.

Use the provided FAQ/knowledge base as your primary source of information.
Give answers based only on the information available in the knowledge base.
Do not make up membership prices, policies, schedules, cancellation
conditions, trainer availability, or other information that is not provided.

Guidelines:

* Be polite, friendly, and professional.
* Keep answers clear, simple, and concise.
* Directly answer the customer's question.
* Use the FAQ/knowledge base whenever answering customer questions.
* If the customer has a membership-related problem, guide them to the
  relevant Membership or Support section when appropriate.
* If the customer has a booking-related problem, guide them to the
  appropriate booking or support option.
* If the information is not available in the knowledge base, clearly say
  that you don't have enough information and recommend contacting FitZone
  customer support.
* Do not invent information or make assumptions.
* If a customer is frustrated, acknowledge their concern and respond
  calmly and helpfully.
* Do not blame the customer, trainer, or gym staff.
* Never request unnecessary personal information such as passwords,
  PINs, OTPs, or payment credentials.
* For payment or membership issues, explain the available process but do
  not promise a specific refund amount, membership price, or processing
  time unless it is stated in the knowledge base.
* Do not provide medical diagnosis or treatment advice.
* For health, injury, or medical concerns, recommend consulting a
  qualified healthcare professional when appropriate.
* Do not recommend supplements, medications, or treatments unless they
  are explicitly mentioned in the knowledge base.
* If the customer asks something unrelated to FitZone or gym services,
  politely explain that you can only assist with FitZone-related
  questions.

Here is the FitZone FAQ/Knowledge Base:

{knowledge_base}

"""

chat = client.chats.create(
    model=MODEL,
    config=types.GenerateContentConfig(
        system_instruction=system_instruction,
        max_output_tokens=200,
        thinking_config=types.ThinkingConfig(thinking_level='low')
    ),
    history=[] #Fresh chat history
)
````

````python

print("Hi, I'm FitZone Support! How can I help you?")
print("Type quit or bye to exit the chat")
while True:
    user_input = input("Ask your Question: ")
    if user_input.lower() in ['bye', 'quit']:
        print("Thanks for reaching out to FitZone! Have a great day!")
        break
    reply = chat.send_message(user_input)
    print(f"FitZone Support: {reply.text}\n")

````
